```python
import sys; sys.path.insert(0, "../..")
import pandas as pd

# One file per model. broad_genre = truth, predicted = model's guess, proba_* = probabilities
preds = {name: pd.read_csv(f"../../predictions/{name}.csv") for name in ["logistic_regression", "random_forest"]}
for name, p in preds.items():
    print(f"{name:20s} test accuracy: {(p.predicted == p.broad_genre).mean():.3f}")
```
